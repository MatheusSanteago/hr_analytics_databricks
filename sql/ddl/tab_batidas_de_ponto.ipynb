```python
%sql
CREATE TABLE IF NOT EXISTS projetos.silver.tab_batidas_de_ponto (
  id_batida              STRING       NOT NULL COMMENT 'Identificador único da batida (ADP)',
  id_cracha              STRING                COMMENT 'Crachá do colaborador no ADP',
  worker_id              STRING                COMMENT 'Colaborador (Workday)',
  data_batida            DATE                  COMMENT 'Data local da batida',
  hora_batida            STRING                COMMENT 'Hora local da batida (HH:MM:SS)',
  data_hora_utc          TIMESTAMP             COMMENT 'Instante da batida em UTC',
  tipo_batida            STRING                COMMENT 'Entrada, Saida_Almoco, Retorno_Almoco, Saida',
  sequencia_dia          STRING                COMMENT 'Ordem informada pelo relógio; pode estar fora de ordem',
  origem                 STRING                COMMENT 'Relogio, App Mobile, Web, Ajuste Manual',
  equipamento_id         STRING                COMMENT 'Equipamento (REP) que registrou a batida',
  localizacao            STRING                COMMENT 'Escritório, Remoto ou Fora do perímetro',
  latitude               STRING                COMMENT 'Latitude da batida (texto, como vem da origem)',
  longitude              STRING                COMMENT 'Longitude da batida (texto, como vem da origem)',
  status                 STRING                COMMENT 'Valida, Pendente de Aprovacao, Ajustada',
  justificativa          STRING                COMMENT 'Justificativa da batida, quando houver',
  fuso_horario           STRING                COMMENT 'Fuso horário IANA do local da batida',
  source_file            STRING                COMMENT 'Metadado de ingestão: arquivo de origem',
  file_modification_time TIMESTAMP             COMMENT 'Metadado de ingestão: data de modificação do arquivo',
  CONSTRAINT pk_batidas_ponto PRIMARY KEY (id_batida) NOT ENFORCED
)
USING DELTA
COMMENT 'Silver: batidas de ponto do ADP, ingeridas do Google Drive'
TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'true',
  'delta.enableChangeDataFeed'       = 'true'
);
```
