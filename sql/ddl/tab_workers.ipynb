```python
%sql
CREATE OR REPLACE TABLE projetos.silver.tab_workers (
  Worker_ID                STRING         NOT NULL COMMENT 'Identificador do colaborador (Workday)',
  Worker_Type              STRING                  COMMENT 'Employee ou Contingent Worker',
  First_Name               STRING                  COMMENT 'Primeiro nome',
  Last_Name                STRING                  COMMENT 'Sobrenome',
  Legal_Name               STRING                  COMMENT 'Nome legal completo',
  Email_Work               STRING                  COMMENT 'E-mail corporativo',
  Phone_Number             STRING                  COMMENT 'Telefone com DDI (texto, preserva o +)',
  Gender                   STRING                  COMMENT 'Male, Female ou Not Declared',
  Date_of_Birth            DATE                    COMMENT 'Data de nascimento (dado pessoal)',
  National_ID_Type         STRING                  COMMENT 'Tipo do documento nacional (SSN, CPF, Steuer-ID etc.)',
  National_ID              STRING                  COMMENT 'Número do documento nacional (texto, preserva zeros à esquerda)',
  Country                  STRING                  COMMENT 'Código do país (BR, US, DE, GB, PT, MX)',
  City                     STRING                  COMMENT 'Cidade',
  Location                 STRING                  COMMENT 'Escritório de lotação',
  Company                  STRING                  COMMENT 'Empresa legal',
  Cost_Center              STRING                  COMMENT 'Centro de custo',
  Supervisory_Organization STRING                  COMMENT 'Organização de supervisão',
  Department               STRING                  COMMENT 'Departamento',
  Job_Profile              STRING                  COMMENT 'Perfil de cargo',
  Business_Title           STRING                  COMMENT 'Cargo exibido',
  Job_Family               STRING                  COMMENT 'Família de cargos',
  Management_Level         STRING                  COMMENT 'Nível de gestão',
  Manager_ID               STRING                  COMMENT 'Worker_ID do gestor direto (nulo apenas para o CEO)',
  Worker_Status            STRING                  COMMENT 'Active, On Leave ou Terminated',
  Employee_Type            STRING                  COMMENT 'Regular ou tipo de contrato',
  Time_Type                STRING                  COMMENT 'Full time ou Part time',
  Pay_Rate_Type            STRING                  COMMENT 'Salary ou Hourly',
  Scheduled_Weekly_Hours   DECIMAL(5,2)            COMMENT 'Jornada semanal contratada (horas)',
  Currency                 STRING                  COMMENT 'Moeda do salário (BRL, EUR, GBP, MXN, USD)',
  Base_Pay_Annual          DECIMAL(18,2)           COMMENT 'Salário base anual, na moeda do colaborador',
  Original_Hire_Date       DATE                    COMMENT 'Data da primeira contratação',
  Hire_Date                DATE                    COMMENT 'Data da contratação atual',
  Termination_Date         DATE                    COMMENT 'Data de desligamento; nula se não desligado',
  Termination_Reason       STRING                  COMMENT 'Motivo do desligamento',
  Time_Zone                STRING                  COMMENT 'Fuso horário IANA do colaborador',
  ADP_Badge_ID             STRING                  COMMENT 'Crachá no ADP, usado para ligar às batidas',
  source_file                STRING  COMMENT 'Metadado do nome do arquivo de origem',
  file_modification_time TIMESTAMP COMMENT 'Metadado do arquivo de origem (usado no notebook)',
  CONSTRAINT pk_workers PRIMARY KEY (Worker_ID) NOT ENFORCED
)
USING DELTA
COMMENT 'Silver: colaboradores (Workday / Fake), ingeridos do Google Drive'
TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'true',
  'delta.enableChangeDataFeed'       = 'true'
);
```
