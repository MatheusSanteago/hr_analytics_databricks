```python
%sql
CREATE TABLE IF NOT EXISTS projetos.silver.tab_eventos_colaboradores (
  event_id                      STRING        NOT NULL COMMENT 'Identificador único do evento (Workday)',
  worker_id                     STRING                 COMMENT 'Colaborador afetado pelo evento',
  business_process              STRING                 COMMENT 'Hire, Promotion, Termination, Compensation Change etc.',
  event_reason                  STRING                 COMMENT 'Motivo do evento, ex.: Hire > Backfill',
  effective_date                DATE                   COMMENT 'Data em que o evento passa a valer',
  entered_on                    TIMESTAMP              COMMENT 'Data e hora em que o evento foi registrado (UTC)',
  status                        STRING                 COMMENT 'Successfully Completed, In Progress, Denied, Rescinded',
  initiated_by                  STRING                 COMMENT 'Worker_ID de quem iniciou o evento',
  completed_on                  TIMESTAMP              COMMENT 'Data e hora de conclusão (UTC); nulo se em andamento',
  from_job_profile              STRING                 COMMENT 'Cargo antes do evento',
  to_job_profile                STRING                 COMMENT 'Cargo depois do evento',
  from_supervisory_organization STRING                 COMMENT 'Organização de supervisão antes do evento',
  to_supervisory_organization   STRING                 COMMENT 'Organização de supervisão depois do evento',
  from_location                 STRING                 COMMENT 'Local antes do evento',
  to_location                   STRING                 COMMENT 'Local depois do evento',
  from_base_pay                 DECIMAL(18,2)          COMMENT 'Salário base antes do evento, na moeda do evento',
  to_base_pay                   DECIMAL(18,2)          COMMENT 'Salário base depois do evento, na moeda do evento',
  currency                      STRING                 COMMENT 'Moeda do salário (BRL, EUR, GBP, MXN, USD)',
  comment                       STRING                 COMMENT 'Comentário livre do evento',
  source_file                   STRING                 COMMENT 'Metadado de ingestão: arquivo de origem',
  file_modification_time        TIMESTAMP              COMMENT 'Metadado de ingestão: data de modificação do arquivo',
  CONSTRAINT pk_eventos_colaboradores PRIMARY KEY (event_id) NOT ENFORCED
)
USING DELTA
COMMENT 'Silver: eventos de ciclo de vida do colaborador (Workday), ingeridos do Google Drive'
TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'true',
  'delta.enableChangeDataFeed'       = 'true'
);
```
